Analyze Survival patterns 

```python
# survival Rate
survival_rate = int(titanic["Survived"].mean() *100)
print(survival_rate)
```

```python
titanic.info()
```

```python
survival_rate_male = titanic[titanic['Sex'] == 'male']['Survived'].mean()*100
print(survival_rate_male)
survival_rate_female = titanic[titanic['Sex']=='female']["Survived"].mean()*100
print(survival_rate_female)
print(survival_rate_female+survival_rate_male)
```

finding the servival rate of children age < 16

```python
print(titanic[titanic["Age"]<16]["Survived"].mean()*100)
```

```python
titanic[titanic['Survived']==1 & titanic['Age'].isna()].sort_values(by='Age',ascending=True).min()
```

16 < Age < 60 

```python
titanic[(titanic['Age']>16) &( titanic['Age']<60)].sort_values(by='Age',ascending=True)
```

Creating an Age group based on the child Adult and Senior 


```python
bins = [0,16,60 ,float('inf')]
labels = ['child',"Adult",'senior']
titanic['AgeGroup'] = pd.cut(titanic['Age'],bins=bins,labels=labels,right=False)
titanic.tail()
```

```python
titanic[titanic['AgeGroup']=='child']
```

```python
titanic.groupby('AgeGroup')['Survived'].mean()*100
```
