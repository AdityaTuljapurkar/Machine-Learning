create a matrix containing the marks of 3 students in five subject and find the 
1 > Avg of all marks 
2> find the max marks in each subject 

```python
import numpy as np 
from datetime import datetime 

print(datetime.now())
marks = np.array([[30,45,29],[43,35,42],[22,43,24],[0,5,43],[43,35,36]])

mean_scores = np.mean(marks,axis=1,dtype=int)

max_preson_score = np.max(marks, axis=0)

person_max_score =  np.argmax(marks,axis=0)
overall_topper =  np.argmax(np.sum(marks,axis=0),axis=0)

print(datetime.now())
print ("Avg score : ",mean_scores)
print("Max score : ", max_preson_score)
print("person_max_score : ", person_max_score)
print("overall_topper : ",overall_topper)
```
