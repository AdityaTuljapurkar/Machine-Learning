```python
import matplotlib.pyplot as plt 

epoches = [1,2,3,4,5]
train_loss = [0.9,0.65,0.6,0.7,0.34]
val_loss = [1.0,0.7,0.65,0,0.56]
```

```python
plt.plot(epoches,train_loss,label = 'training loss')
plt.plot(epoches,val_loss , label = 'value loss')
plt.xlabel('epoches')
plt.ylabel('loss')
plt.legend()
plt.grid(True)
plt.title('loss of values')
plt.show()
```

```python
import seaborn as sas 
import matplotlib.pyplot as plt 
import plotly.express as ply
data = sas.load_dataset('titanic')
```

```python
# data.drop(columns=['deck'],inplace=True)
data['age'] = data['age'].fillna(data['age'].mean())
data['embark_town'] = data['embark_town'].fillna(data['embark_town'].mode()[0] ,)
# data.info()
print(data['age'].isna().sum())
print(data['embark_town'].isna().sum())

data.info()
```

```python
import plotly.express as px
import pandas as pd
data['class_age'] = pd.cut(
    data['age'],
    bins=[0,18,25,55,80],
    labels = ['child','young','mid','old']
)
data['survival_percent'] = data['survived'] * 100

fig2 = px.histogram(
    data,
    x='embark_town',
    y='survival_percent',
    color='sex',
    facet_col='class_age',
    histfunc='avg'
    


)

fig2.show()
```
